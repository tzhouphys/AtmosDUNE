```python
COLORS_SONG = [
    '#3b3e37',           
    '#535e4e',           
    '#6d593d',
    '#9c8559',
    '#d2c591',
]

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, color=COLORS_SONG[0])
plt.plot(x, np.cos(x), color=COLORS_SONG[1])
plt.plot(x, np.tan(x), color=COLORS_SONG[2])
plt.plot(x, np.exp(-x), color=COLORS_SONG[3])
plt.plot(x, np.log(x + 1), color=COLORS_SONG[4])
plt.show()
```
